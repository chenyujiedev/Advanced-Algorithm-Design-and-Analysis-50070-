```python
# Langchain을 거치지 않고 네이티브 ollama 라이브러리를 직접 호출하면 더 가볍습니다
import ollama
```

```python
# 문제를 미리 정의해 두면 재사용하기 편리하고, 매번 채팅창에 하드코딩할 필요가 없습니다
prompt="今天上海的天气怎么样？"
```

```python
# ollama.chat을 사용하여 모델을 직접 호출할 때, messages 형식은 OpenAI와 동일하며 role + content 형식입니다.
response=ollama.chat(
    model="llama3.1",
    messages=[
        {
        "role":"user",
        "content":prompt

    }]
)
```

```python
# 반환된 전체 response 객체를 확인해 보세요
response
```

```python
# 본문 내용만 추출하고, 그 많은 메타데이터는 제외해 주세요. 이것이야말로 제가 정말로 원하는 답변입니다.
response["message"]["content"]
```
